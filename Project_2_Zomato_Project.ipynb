## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Merge datasets
df = pd.merge(restaurant_df, review_df,
              left_on='Name',
              right_on='Restaurant',
              how='inner')

# Drop duplicate column and rename for consistency
df.drop('Restaurant', axis=1, inplace=True)
df.rename(columns={'Name': 'Restaurant'}, inplace=True)

# Check missing values
df.isnull().sum()

# Drop rows with missing important values
df.dropna(subset=['Rating', 'Cost'], inplace=True)

# Fill missing review text
df['Review'].fillna('', inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Clean and convert Cost

df['Cost'] = df['Cost'].astype(str)

df['Cost'] = df['Cost'].str.replace(',', '', regex=True)
df['Cost'] = df['Cost'].str.replace('₹', '', regex=True)

# Extract numeric part
df['Cost'] = df['Cost'].str.extract('(\d+)')

# Convert to numeric
df['Cost'] = pd.to_numeric(df['Cost'], errors='coerce')

# Drop rows where conversion failed
df.dropna(subset=['Cost'], inplace=True)

# Remove outliers using IQR

Q1 = df['Cost'].quantile(0.25)
Q3 = df['Cost'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['Cost'] >= Q1 - 1.5 * IQR) &
        (df['Cost'] <= Q3 + 1.5 * IQR)]

# Select relevant columns

df = df[['Restaurant', 'Cost', 'Cuisines', 'Collections', 'Rating', 'Review']]

# Final check
df.head()
df.shape
````

### What all manipulations have you done and insights you found?

In this step the restaurant. The review dataset were merged to create a unified dataset for analysis. So I removed rows that were missing information like ratings and cost. For text fields that were missing information I just left them empty so I do not lose any information. We also removed duplicate records because I do not want the information twice. This helps make the data better.

I had to do something about outliers in numbers like cost.I used the IQR method to deal with values that could mess up our analysis and clustering. Finally I only kept the columns that're important. ### 4. Textual Data Preprocessing
(It's mandatory for textual dataset i.e., NLP, Sentiment Analysis, Text Clustering etc.)

#### 1. Expand Contraction

````python
# Expand Contraction
df['Review'] = df['Review'].str.replace("n't", " not")
````

#### 2. Lower Casing

````python
# Lower Casing
df['Review'] = df['Review'].str.lower()
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
import string
df['Review'] = df['Review'].str.translate(str.maketrans('', '', string.punctuation))
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits contain digits
import re
df['Review'] = df['Review'].apply(lambda x: re.sub(r'http\S+|www\S+', '', x))
df['Review'] = df['Review'].apply(lambda x: re.sub(r'\d+', '', x))
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

df['Review'] = df['Review'].apply(
    lambda x: ' '.join([word for word in x.split() if word not in stop_words])
)

df['Review'] = df['Review'].str.strip()
````

#### 6. Rephrase Text

````python
# Rephrase Text
df['Clean_Review'] = df['Review']
````

#### 7. Tokenization

````python
# Tokenization
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

df['Tokens'] = df['Review'].apply(word_tokenize)
````

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

df['Tokens'] = df['Tokens'].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)
````

##### Which text normalization technique have you used and why?

Lemmatization was used because it converts words into their base form while preserving meaning, which is more accurate than stemming.

#### 9. Part of speech tagging

````python
# POS Taging
import nltk

nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

from nltk import pos_tag

df['POS_Tags'] = df['Tokens'].apply(pos_tag)
````

### ML Model - 1

````python
# ML Model - 1 Implementation
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Elbow method
inertia = []

for k in range(2, 10):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Plot Elbow
plt.plot(range(2,10), inertia, marker='o')
plt.title("Elbow Method")
plt.xlabel("K")
plt.ylabel("Inertia")
plt.show()

# Fit the Algorithm
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_scaled)

df['Cluster_KMeans'] = labels

# Predict on the model
score = silhouette_score(X_scaled, labels)
print("Silhouette Score:", score)
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Visualizing evaluation Metric Score chart
scores = []

for k in range(2,10):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(X_scaled)
    scores.append(silhouette_score(X_scaled, labels))

plt.plot(range(2,10), scores, marker='o')
plt.title("Silhouette Score vs K")
plt.xlabel("K")
plt.ylabel("Score")
plt.show()
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
best_score = -1
best_k = 2

for k in range(2,10):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)

    if score > best_score:
        best_score = score
        best_k = k

print("Best K:", best_k)
````
